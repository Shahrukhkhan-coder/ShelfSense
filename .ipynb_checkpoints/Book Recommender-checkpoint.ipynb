```python
import numpy as np 
import pandas as pd
```

```python
books = pd.read_csv('books.csv')
```

```python
books.columns
```

```python
books = books[['ISBN', 'Book-Title', 'Book-Author', 'Year-Of-Publication', 'Publisher']]
```

```python
books.head(3)
```

```python
books.rename(columns={'Book-Title':'title', 'Book-Author':'author', 'Year-Of-Publication':'year', 'Publisher':'publisher'},inplace=True)
```

```python
books.head(3)
```

```python
users = pd.read_csv('Users.csv')
```

```python
users.head(3)
```

```python
users.rename(columns={'User-ID':'user_id', 'Location':'location', 'Age':'age'},inplace=True)
```

```python
users.head(3)
```

```python
ratings = pd.read_csv('Ratings.csv')
```

```python
ratings.head(3)
```

```python
ratings.rename(columns={'User-ID':'user_id', 'Book-Rating':'rating'},inplace=True)
```

```python
ratings.head(3)
```

```python
books.shape
```

```python
users.shape
```

```python
ratings.shape
```

```python
x = ratings['user_id'].value_counts() > 200
```

```python
y = x[x].index
```

```python
ratings = ratings[ratings['user_id'].isin(y)]
```

```python
ratings.shape
```

```python
ratings_with_books = ratings.merge(books, on='ISBN')
```

```python
ratings_with_books.head(3)
```

```python
ratings_with_books.shape
```

```python
number_rating = ratings_with_books.groupby('title')['rating'].count().reset_index()
```

```python
number_rating.rename(columns={'rating':'number of rating'},inplace=True)
```

```python
number_rating.head(3)
```

```python
final_rating = ratings_with_books.merge(number_rating, on='title')
```

```python
final_rating.head(3)
```

```python
final_rating.shape
```

```python
final_rating = final_rating[final_rating['number of rating'] >= 50]
```

```python
final_rating.shape
```

```python
final_rating.drop_duplicates(['user_id','title'], inplace=True)
```

```python
final_rating.shape
```

```python
book_pivot = final_rating.pivot_table(columns='user_id', index='title', values='rating')
```

```python
book_pivot.head(3)
```

```python
book_pivot.shape
```

```python
book_pivot.fillna(0, inplace=True)
```

```python
book_pivot.head(3)
```

```python
from scipy.sparse import csr_matrix

book_sparse = csr_matrix(book_pivot)
```

```python
type(book_sparse)
```

```python
from sklearn.neighbors import NearestNeighbors

model = NearestNeighbors(algorithm='brute')
```

```python
model.fit(book_sparse)
```

```python
distances, suggestions = model.kneighbors(book_pivot.iloc[237, :].values.reshape(1,-1), n_neighbors=6)
```

```python
suggestions
```

```python
for i in range(len(suggestions)):
    print(book_pivot.index[suggestions[i]])
```

```python
int(np.where(book_pivot.index == 'Animal Farm')[0][0])
```

```python
def recommend_book(book_name):
    book_id = int(np.where(book_pivot.index == book_name)[0][0])
    distances, suggestions = model.kneighbors(book_pivot.iloc[book_id, :].values.reshape(1,-1), n_neighbors=6)

    for i in range(len(suggestions[0])):
        if i == 0:
            continue
        else:
            print(book_pivot.index[suggestions[0][i]])
```

```python
recommend_book('Harry Potter and the Chamber of Secrets (Book 2)')
```
